```python
df_clean = df.drop_duplicates().copy()

print("Original shape:", df.shape)
print("After removing duplicates:", df_clean.shape)
print("Duplicates remaining:", df_clean.duplicated().sum())
```

```python
X = df_clean.drop("Class", axis=1)
y = df_clean["Class"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_temp, X_test, y_temp, y_test = train_test_split(
    X,
    y,
    test_size=0.10,
    stratify=y,
    random_state=42
)

print("Temporary train shape:", X_temp.shape)
print("Test shape:", X_test.shape)

print("\nTest class distribution:")
print(y_test.value_counts())
```

```python
X_train, X_val, y_train, y_val = train_test_split(
    X_temp,
    y_temp,
    test_size=0.111111,
    stratify=y_temp,
    random_state=42
)

print("Train shape:", X_train.shape)
print("Validation shape:", X_val.shape)
print("Test shape:", X_test.shape)
```

```python
print("Train:")
print(y_train.value_counts())
print(y_train.value_counts(normalize=True) * 100)

print("\nValidation:")
print(y_val.value_counts())
print(y_val.value_counts(normalize=True) * 100)

print("\nTest:")
print(y_test.value_counts())
print(y_test.value_counts(normalize=True) * 100)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = X_train.copy()
X_val_scaled = X_val.copy()
X_test_scaled = X_test.copy()

X_train_scaled[["Time", "Amount"]] = scaler.fit_transform(
    X_train[["Time", "Amount"]]
)

X_val_scaled[["Time", "Amount"]] = scaler.transform(
    X_val[["Time", "Amount"]]
)

X_test_scaled[["Time", "Amount"]] = scaler.transform(
    X_test[["Time", "Amount"]]
)
```

```python
print("Before scaling:")
print(X_train[["Time", "Amount"]].describe())

print("\nAfter scaling:")
print(X_train_scaled[["Time", "Amount"]].describe())
```

### Class Weighted Logistic Regression

```python
from sklearn.linear_model import LogisticRegression

weighted_model = LogisticRegression(
    max_iter=1000,
    class_weight="balanced",
    random_state=42
)

weighted_model.fit(X_train_scaled, y_train)

print("Class-weighted model trained successfully.")
```

```python
y_val_pred_weighted = weighted_model.predict(X_val_scaled)

y_val_proba_weighted = weighted_model.predict_proba(
    X_val_scaled
)[:, 1]
```

```python
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    roc_auc_score,
    average_precision_score,
    accuracy_score
)

print("Confusion Matrix:")
print(confusion_matrix(y_val, y_val_pred_weighted))

print("\nClassification Report:")
print(
    classification_report(
        y_val,
        y_val_pred_weighted,
        target_names=["Legitimate", "Fraud"]
    )
)

roc_auc = roc_auc_score(
    y_val,
    y_val_proba_weighted
)

pr_auc = average_precision_score(
    y_val,
    y_val_proba_weighted
)

accuracy = accuracy_score(
    y_val,
    y_val_pred_weighted
)

print("ROC-AUC:", roc_auc)
print("PR-AUC:", pr_auc)
print("Accuracy:", accuracy)
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score

thresholds = [0.10, 0.20, 0.30, 0.40, 0.50, 0.60, 0.70, 0.80, 0.90]

for threshold in thresholds:

    y_pred_threshold = (
        y_val_proba >= threshold
    ).astype(int)

    precision = precision_score(
        y_val,
        y_pred_threshold,
        zero_division=0
    )

    recall = recall_score(
        y_val,
        y_pred_threshold,
        zero_division=0
    )

    f1 = f1_score(
        y_val,
        y_pred_threshold,
        zero_division=0
    )

    print(
        f"Threshold: {threshold:.2f} | "
        f"Precision: {precision:.3f} | "
        f"Recall: {recall:.3f} | "
        f"F1: {f1:.3f}"
    )
```

```python
from sklearn.metrics import precision_recall_curve

precision, recall, thresholds = precision_recall_curve(
    y_val,
    y_val_proba
)

print("Number of thresholds:", len(thresholds))
```

```python
f1_scores = (
    2 * precision[:-1] * recall[:-1]
    / (precision[:-1] + recall[:-1] + 1e-10)
)

best_index = f1_scores.argmax()

best_threshold = thresholds[best_index]
best_precision = precision[best_index]
best_recall = recall[best_index]
best_f1 = f1_scores[best_index]

print("Best threshold:", best_threshold)
print("Precision:", best_precision)
print("Recall:", best_recall)
print("F1-score:", best_f1)
```

```python
plt.figure(figsize=(8, 6))

plt.plot(recall, precision)

plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title("Precision-Recall Curve")

plt.grid()

plt.show()
```

```python
plt.figure(figsize=(10, 5))

plt.plot(thresholds, f1_scores)

plt.axvline(
    best_threshold,
    linestyle="--",
    label=f"Best threshold = {best_threshold:.3f}"
)

plt.xlabel("Threshold")
plt.ylabel("F1-score")
plt.title("F1-score vs Classification Threshold")

plt.legend()
plt.grid()

plt.show()
```

```python
y_val_best_pred = (
    y_val_proba >= best_threshold
).astype(int)
```

```python
from sklearn.metrics import classification_report, confusion_matrix

print("Confusion Matrix:")
print(confusion_matrix(y_val, y_val_best_pred))

print("\nClassification Report:")
print(
    classification_report(
        y_val,
        y_val_best_pred,
        target_names=["Legitimate", "Fraud"]
    )
)
```

### Random Forest

```python
from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(
    n_estimators=200,
    max_depth=None,
    min_samples_leaf=2,
    class_weight=None,
    random_state=42,
    n_jobs=-1
)

rf_model.fit(X_train_scaled, y_train)

print("Random Forest trained successfully.")
```

```python
y_val_pred_rf = rf_model.predict(X_val_scaled)

y_val_proba_rf = rf_model.predict_proba(
    X_val_scaled
)[:, 1]
```

```python
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    roc_auc_score,
    average_precision_score,
    accuracy_score
)

print("Confusion Matrix:")
print(confusion_matrix(y_val, y_val_pred_rf))

print("\nClassification Report:")
print(
    classification_report(
        y_val,
        y_val_pred_rf,
        target_names=["Legitimate", "Fraud"]
    )
)

print("ROC-AUC:",
      roc_auc_score(y_val, y_val_proba_rf))

print("PR-AUC:",
      average_precision_score(y_val, y_val_proba_rf))

print("Accuracy:",
      accuracy_score(y_val, y_val_pred_rf))
```

```python
### Random Forest Threshold Tuning
precision_rf, recall_rf, thresholds_rf = precision_recall_curve(
    y_val,
    y_val_proba_rf
)

f1_scores_rf = (
    2 * precision_rf[:-1] * recall_rf[:-1]
    / (precision_rf[:-1] + recall_rf[:-1] + 1e-10)
)

best_index_rf = f1_scores_rf.argmax()

best_threshold_rf = thresholds_rf[best_index_rf]
best_precision_rf = precision_rf[best_index_rf]
best_recall_rf = recall_rf[best_index_rf]
best_f1_rf = f1_scores_rf[best_index_rf]

print("Best threshold:", best_threshold_rf)
print("Precision:", best_precision_rf)
print("Recall:", best_recall_rf)
print("F1-score:", best_f1_rf)
```

```python
y_val_best_pred_rf = (
    y_val_proba_rf >= best_threshold_rf
).astype(int)
```

```python
print("Confusion Matrix:")
print(confusion_matrix(y_val, y_val_best_pred_rf))

print("\nClassification Report:")
print(
    classification_report(
        y_val,
        y_val_best_pred_rf,
        target_names=["Legitimate", "Fraud"]
    )
)
```

### SMOTE

```python
from imblearn.over_sampling import SMOTE

smote = SMOTE(
    sampling_strategy=0.1,
    random_state=42
)

X_train_smote, y_train_smote = smote.fit_resample(
    X_train_scaled,
    y_train
)

print("Original training shape:", X_train_scaled.shape)
print("After SMOTE:", X_train_smote.shape)

print("\nOriginal class distribution:")
print(y_train.value_counts())

print("\nAfter SMOTE:")
print(y_train_smote.value_counts())
```

```python
rf_smote = RandomForestClassifier(
    n_estimators=200,
    min_samples_leaf=2,
    random_state=42,
    n_jobs=-1
)

rf_smote.fit(X_train_smote, y_train_smote)

y_val_proba_smote = rf_smote.predict_proba(
    X_val_scaled
)[:, 1]

y_val_pred_smote = (
    y_val_proba_smote >= 0.5
).astype(int)
```

```python
print("Confusion Matrix:")
print(confusion_matrix(y_val, y_val_pred_smote))

print("\nClassification Report:")
print(
    classification_report(
        y_val,
        y_val_pred_smote,
        target_names=["Legitimate", "Fraud"]
    )
)

print("ROC-AUC:",
      roc_auc_score(y_val, y_val_proba_smote))

print("PR-AUC:",
      average_precision_score(y_val, y_val_proba_smote))
```

```python
precision_smote, recall_smote, thresholds_smote = precision_recall_curve(
    y_val,
    y_val_proba_smote
)

f1_scores_smote = (
    2 * precision_smote[:-1] * recall_smote[:-1]
    / (
        precision_smote[:-1]
        + recall_smote[:-1]
        + 1e-10
    )
)

best_index_smote = f1_scores_smote.argmax()

best_threshold_smote = thresholds_smote[best_index_smote]
best_precision_smote = precision_smote[best_index_smote]
best_recall_smote = recall_smote[best_index_smote]
best_f1_smote = f1_scores_smote[best_index_smote]

print("Best threshold:", best_threshold_smote)
print("Precision:", best_precision_smote)
print("Recall:", best_recall_smote)
print("F1-score:", best_f1_smote)
```

```python
y_val_best_pred_smote = (
    y_val_proba_smote >= best_threshold_smote
).astype(int)

print("Confusion Matrix:")
print(confusion_matrix(y_val, y_val_best_pred_smote))

print("\nClassification Report:")
print(
    classification_report(
        y_val,
        y_val_best_pred_smote,
        target_names=["Legitimate", "Fraud"]
    )
)
```

```python
# Final test probabilities
y_test_proba_final = rf_smote.predict_proba(
    X_test_scaled
)[:, 1]

# Apply threshold selected using validation data
y_test_pred_final = (
    y_test_proba_final >= best_threshold_smote
).astype(int)

print("Final Test Confusion Matrix:")
print(confusion_matrix(y_test, y_test_pred_final))

print("\nFinal Test Classification Report:")
print(
    classification_report(
        y_test,
        y_test_pred_final,
        target_names=["Legitimate", "Fraud"]
    )
)

print(
    "Final Test ROC-AUC:",
    roc_auc_score(y_test, y_test_proba_final)
)

print(
    "Final Test PR-AUC:",
    average_precision_score(y_test, y_test_proba_final)
)
```
